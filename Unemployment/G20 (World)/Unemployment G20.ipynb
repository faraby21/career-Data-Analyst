```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df_main = pd.read_csv('API_SL.UEM.TOTL.ZS_DS2_en_csv_v2_33398.csv', skiprows=4)
df_country = pd.read_csv('Metadata_Country_API_SL.UEM.TOTL.ZS_DS2_en_csv_v2_33398.csv')
df_indicator = pd.read_csv('Metadata_Indicator_API_SL.UEM.TOTL.ZS_DS2_en_csv_v2_33398.csv')
```

```python
df_main.head()
```

```python
tahun_cols = [str(year) for year in range(1960, 2023)]

df_long = df_main.melt(
    id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'],
    value_vars=tahun_cols,
    var_name='Year',
    value_name='Unemployment_Rate'
)


df_long['Year'] = pd.to_numeric(df_long['Year'])

list_Country = ['IDN', 'MYS', 'SGP', 'THAI', 'LAO', 'VNM', 'PHL', 'KHM']
df_SoutheastAsia_Country = df_long[df_long['Country Code'].isin(list_Country)]

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_SoutheastAsia_Country, 
    x='Year', 
    y='Unemployment_Rate', 
    hue='Country Name', 
    marker='o'
)


plt.title('Comparison of Unemployment Rates Across Several Countries', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Unemployment Rate (%)', fontsize=12)
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df_long['Year'] = pd.to_numeric(df_long['Year'])


list_negara = ['IDN', 'MYS', 'SGP', 'THAI', 'LAO', 'VNM', 'PHL', 'KHM']


df_bandingan = df_long[
    (df_long['Year'] >= 2020) & 
    (df_long['Year'] <= 2025) & 
    (df_long['Country Code'].isin(list_negara))
]

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_bandingan, 
    x='Year', 
    y='Unemployment_Rate', 
    hue='Country Name', 
    marker='o', 
    linewidth=2.5
)

# 5. Pengaturan Tampilan Grafik
plt.title('Compare Unemployment Rate In South East Asia (2020 - 2025)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Unemployment Rate (%)', fontsize=12)
plt.xticks(range(2020, 2026)) # Memastikan sumbu X menampilkan tahun secara utuh dari 2020-2025
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

lastly_year = '2024'
df_2024 = df_long[df_long['Year'] == 2024].dropna(subset=['Unemployment_Rate'])
df_2024 = df_2024[df_2024['Unemployment_Rate'] > 0] 


top_5_country = df_2024.sort_values(by='Unemployment_Rate', ascending=False).head(5)


print(top_5_country[['Country Name', 'Unemployment_Rate']])

plt.figure(figsize=(10, 5))
sns.barplot(data=top_5_country, x='Unemployment_Rate', y='Country Name', palette='viridis')
plt.title('5 Country with highest rate unemployment (2024)')
plt.xlabel('Unemployment Rate (%)')
plt.show()
```
