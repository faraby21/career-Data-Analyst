```python
df_bersih = pd.read_csv(
    'Tingkat Pengangguran Terbuka Berdasarkan Tingkat Pendidikan, 2025.csv', 
    skiprows=3, names=['Tingkat Pendidikan' , 'Persentase'],
    encoding='utf-8-sig')

print(df_bersih)
```

```python
plt.figure(figsize=(15, 10))

ax = plt.gca()
df_bersih.plot(kind='bar', x='Tingkat Pendidikan', y='Persentase', ax=ax, color='#0077B6', edgecolor= 'Black')
plt.title('Tingkat Pengangguran Terbuka Berdasarkan Pendidikan (2025)',fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tingkat Pendidikan', fontsize=12 )
plt.ylabel('Persentase (%)', fontsize=12)
plt.xticks(rotation=30, ha='right', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 10))

ax = plt.gca()
df_bersih.plot(kind='line', x='Tingkat Pendidikan', y='Persentase', ax=ax, color='#0077B6', linewidth=2, markersize=8)
plt.title('Tren Pengangguran Terbuka Berdasarkan Pendidikan (2025)',fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tingkat Pendidikan', fontsize=12 )
plt.ylabel('Persentase (%)', fontsize=12)
plt.xticks(rotation=30, ha='right', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df_bersih = pd.read_csv(
    'Tingkat Pengangguran Terbuka Berdasarkan Tingkat Pendidikan, 2025.csv',
    skiprows=3,
    names=['Tingkat Pendidikan', 'Persentase'],
    encoding='utf-8-sig'
)

plt.figure(figsize=(15, 10))

plt.pie( df_bersih['Persentase'], 
    labels=df_bersih['Tingkat Pendidikan'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#0077B6', '#0096C7', '#03045E', '#90E0EF', '#00b4d8', '#caf0f8'],
    textprops={'fontsize': 11})

plt.title('Proporsi Tingkat  Pendidikan') 
plt.tight_layout()
plt.show()
```
