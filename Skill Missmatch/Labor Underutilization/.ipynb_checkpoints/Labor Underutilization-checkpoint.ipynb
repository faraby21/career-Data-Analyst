```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_country = pd.read_csv('underutil_workage_by_country_year.csv')
df_sex = pd.read_csv('underutil_by_pob.csv')
df_edu_pob = pd.read_csv('underutil_by_pob_edu.csv')
df_edu = pd.read_csv('underutil_by_edu.csv')
df_age = pd.read_csv('underutil_by_ageg.csv')
```

```python
#1
negara_penting = ['AT', 'DE', 'FR', 'IT', 'ES'] 
data_panel = df_country[df_country['COUNTRY'].isin(negara_penting)]

g = sns.FacetGrid(
    data_panel,
    col='COUNTRY',       
    col_wrap=3,          
    height=4,            
    aspect=1.5           
)

g.map(plt.plot, 'YEAR', 'underutil_rate', marker='o', linestyle='-', color='#0077B6')

g.set_titles(col_template="Country: {col_name}")
g.set_axis_labels("Year", "Underutil Rate")
g.fig.suptitle('Tren Underutilization Rate Based On Country', fontsize=16, fontweight='bold', y=1.05) 

plt.tight_layout()


plt.show()
```

```python
#2
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df_sex,
    x='POB',
    y='underutil_rate',
    hue='SEX',
    palette=['#0077B6', '#ff006e'],  
    edgecolor='black',
)

plt.title(
    'Comparison Underutilization Rate Based On POB & Gender (SEX)',
    fontsize=13,
    fontweight='bold',
    pad=15,
)
plt.xlabel('POB Category', fontsize=11)
plt.ylabel('Underutil Rate (%)', fontsize=11)

plt.legend(title='Gender (SEX)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
#POB (Place Of Birth)
#0 : for native citizen
#1: Born Abroad (Typically the European Union or neighboring countries) — Immigrants from countries within the immediate region.
#2: Born Abroad (Third Country / outside the region) — Immigrants originating from distant countries or other continents.
#99 : Combined / Total / Unknown Category — Often used as an overall summary or for respondent data where the place of birth is not specifically recorded.
```

```python
#Diagram Based On Education & POB
plt.figure(figsize=(11, 6))

sns.barplot(
    data=df_edu_pob,
    x='POB',
    y='underutil_rate',
    hue='EDU',
    palette='Set2',  
    edgecolor='black',
)

plt.title(
    'Underutilization Rate BBased On Education (EDU) & POB',
    fontsize=13,
    fontweight='bold',
    pad=15,
)
plt.xlabel('POB Category', fontsize=11)
plt.ylabel('Underutil Rate (%)', fontsize=11)

plt.legend(title='Education (EDU)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df_edu,
    x='EDU',
    y='underutil_rate',
    hue='SEX',
    palette=['#0077B6', '#ff006e'], 
    edgecolor='black',
)

plt.title(
    'Underutilization Rate Based On Education Level (EDU) & Gender (SEX)',
    fontsize=13,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Education Level (EDU)', fontsize=11)
plt.ylabel('Underutil Rate (%)', fontsize=11)

plt.legend(title='Gender (SEX)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
#Based On Age
plt.figure(figsize=(11, 6))

sns.barplot(
    data=df_age,
    x='ageg',
    y='underutil_rate',
    hue='EDU',
    palette='viridis',  
    edgecolor='black',
)

plt.title(
    'Underutilization Rate Based On Age & Education Level (EDU)',
    fontsize=13,
    fontweight='bold',
    pad=15,
)
plt.xlabel('Age', fontsize=11)
plt.ylabel('Underutil Rate (%)', fontsize=11)

plt.legend(title='Education Level (EDU)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```
