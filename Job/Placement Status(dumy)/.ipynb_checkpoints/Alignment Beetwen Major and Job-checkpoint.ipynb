```python
df = pd.read_csv('job_placement.csv')

df['placement_status_clean'] = df['placement_status'].astype(str).str.strip().str.title()
df_placed = df[df['placement_status_clean'] == 'Placed']
top_salary = df_placed.sort_values(by='salary', ascending=False).head(5)

print(top_salary[['name', 'college_name', 'stream', 'salary', 'years_of_experience']])
```

```python
#Data Based On employee who have high salary and job
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(data=top_salary, x='salary', color='#3a86ff')

plt.title(
    'Distribution of Graduate Income (Box Plot)',
    fontsize=13,
    fontweight='bold',
    pad=12,
)
plt.xlabel('Salary', fontsize=11)
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

```python
#Diagram average salary who have high income
```

```python
df = pd.read_csv('job_placement.csv')

df['placement_status'] = df['placement_status'].astype(str).str.strip().str.title()
df_Not_placed = df[df['placement_status'] == 'Not Placed']
top_Not_Placed = df_Not_placed.sort_values(by='gpa', ascending=False).head(5)

print(top_Not_Placed[['name', 'college_name', 'stream', 'placement_status', 'gpa', 'years_of_experience']])
```

```python
#Data Based On Who doesnt have job
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=top_Not_Placed, x='gpa', y='name', palette='viridis', hue='name', legend=False)

plt.title('5 Gradute who doesnt have job ', fontsize=14, fontweight='bold')
plt.xlabel('Not Placed', fontsize=12)
plt.ylabel('name graduate', fontsize=12)

plt.grid(axis='x', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
