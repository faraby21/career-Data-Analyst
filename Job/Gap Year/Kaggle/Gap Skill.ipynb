```python
jumlah_data = df['Career Interest'].value_counts()

plt.figure(figsize=(10, 6))
jumlah_data.plot(kind='bar', color='#0077B6', edgecolor='black')

plt.title('Career Interest', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('career choices', fontsize=12)
plt.ylabel('Total', fontsize=12)
plt.xticks(rotation=30, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```
